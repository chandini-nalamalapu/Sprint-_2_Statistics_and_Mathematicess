# 1. Mean

## Definition

Mean is the average value of a group of numbers.

It is calculated by adding all the values and dividing the total by the number of values.

## Formula

Mean = Sum of all values / Number of values

## Numerical Example

Consider the values:

10, 20, 30, 40, 50

Mean = (10 + 20 + 30 + 40 + 50) / 5
Mean = 150 / 5
Mean = 30

So, the mean is 30.

## Business Example

A company wants to find the average monthly sales.

Sales for 5 months are:

10000, 12000, 15000, 13000, 16000

The mean gives the average monthly sales.

## AI/ML Example

In machine learning, mean can be used to understand the average value of a feature.

For example, if the ages of customers are:

20, 25, 30, 35, 40

The mean age is 30.

```python
# Finding the mean using Python

numbers = [10, 20, 30, 40, 50]

total = sum(numbers)
count = len(numbers)

mean = total / count

print("Mean:", mean)
```

```python
import statistics

numbers = [10, 20, 30, 40, 50]

mean = statistics.mean(numbers)

print("Mean:", mean)
```

# 2. Median

## Definition

Median is the middle value of a group of numbers after arranging the values in ascending or descending order.

If there are an even number of values, the median is the average of the two middle values.

## Formula

For odd number of values:

Median = Middle value

For even number of values:

Median = (Two middle values) / 2

## Numerical Example

Consider:

10, 20, 30, 40, 50

The middle value is 30.

So, Median = 30

For even values:

10, 20, 30, 40

Median = (20 + 30) / 2
Median = 25

## Business Example

A company wants to find the typical salary of its employees.

If a few employees have extremely high salaries, the mean can be affected.

Median can give a better idea of the middle salary.

## AI/ML Example

Median can be used to understand the central value of a feature.

It is also useful when handling data containing extreme values or outliers.

```python
# Finding the median

import statistics

numbers = [10, 20, 30, 40, 50]

median = statistics.median(numbers)

print("Median:", median)
```

# 3. Mode

## Definition

Mode is the value that occurs most frequently in a dataset.

A dataset can have one mode or more than one mode.

## Formula

There is no specific mathematical formula for mode.

Mode = Most frequently occurring value

## Numerical Example

Consider:

10, 20, 20, 30, 40

The number 20 occurs two times.

Other values occur only once.

Therefore:

Mode = 20

## Business Example

A retail shop can use mode to find the most commonly purchased product size.

For example:

S, M, M, L, M, S

The most common size is M.

## AI/ML Example

Mode can be used to fill missing values in categorical data.

For example, if a customer dataset has a missing value in the "City" column, the most frequently occurring city can be used as a simple replacement.

```python
# Finding the mode

import statistics

numbers = [10, 20, 20, 30, 40]

mode = statistics.mode(numbers)

print("Mode:", mode)
```

```python
import statistics

cities = ["Hyderabad", "Chennai", "Hyderabad", "Mumbai", "Hyderabad"]

mode = statistics.mode(cities)

print("Most common city:", mode)
```

# 4. Weighted Mean

## Definition

Weighted Mean is an average where different values have different levels of importance or weights.

A value with a higher weight has more influence on the final average.

## Formula

Weighted Mean = Sum of (Value × Weight) / Sum of Weights

## Numerical Example

Suppose a student has:

Assignment = 80, Weight = 20%
Exam = 90, Weight = 80%

Weighted Mean:

= (80 × 0.20) + (90 × 0.80)
= 16 + 72
= 88

So, the weighted mean is 88.

## Business Example

A company can calculate an employee's performance score using different weights.

For example:

Technical Skills = 80, Weight = 60%
Communication = 70, Weight = 40%

Weighted Mean:

= (80 × 0.60) + (70 × 0.40)
= 48 + 28
= 76

## AI/ML Example

Weighted averages can be used when some observations are more important than others.

For example, training data may assign different weights to samples so that important samples have more influence during model training.

# 4. Weighted Mean

## Definition

Weighted Mean is an average where different values have different levels of importance or weights.

A value with a higher weight has more influence on the final average.

## Formula

Weighted Mean = Sum of (Value × Weight) / Sum of Weights

## Numerical Example

Suppose a student has:

Assignment = 80, Weight = 20%
Exam = 90, Weight = 80%

Weighted Mean:

= (80 × 0.20) + (90 × 0.80)
= 16 + 72
= 88

So, the weighted mean is 88.

## Business Example

A company can calculate an employee's performance score using different weights.

For example:

Technical Skills = 80, Weight = 60%
Communication = 70, Weight = 40%

Weighted Mean:

= (80 × 0.60) + (70 × 0.40)
= 48 + 28
= 76

## AI/ML Example

Weighted averages can be used when some observations are more important than others.

For example, training data may assign different weights to samples so that important samples have more influence during model training.

```python
# Finding weighted mean

values = [80, 90]
weights = [0.20, 0.80]

weighted_sum = 0

for value, weight in zip(values, weights):
    weighted_sum = weighted_sum + (value * weight)

weighted_mean = weighted_sum / sum(weights)

print("Weighted Mean:", weighted_mean)
```

# 5. Geometric Mean

## Definition

Geometric Mean is an average that is calculated by multiplying all the values and then taking the nth root, where n is the number of values.

It is mainly useful for values that represent growth rates, ratios, or percentages.

## Formula

Geometric Mean = (x1 × x2 × x3 × ... × xn)^(1/n)

## Numerical Example

Consider:

2, 8

Geometric Mean:

= √(2 × 8)
= √16
= 4

So, the geometric mean is 4.

## Business Example

Geometric mean can be used to calculate the average growth rate of a company over multiple years.

For example, if a business grows by different rates each year, geometric mean can provide an average growth rate over the period.

## AI/ML Example

Geometric mean can be useful when combining ratios or evaluating certain performance measures where relative changes are more meaningful than simple differences.

```python
# Finding geometric mean

import statistics

numbers = [2, 8]

geometric_mean = statistics.geometric_mean(numbers)

print("Geometric Mean:", geometric_mean)
```

```python
import statistics

numbers = [2, 4, 8]

geometric_mean = statistics.geometric_mean(numbers)

print("Geometric Mean:", geometric_mean)
```

# 6. Harmonic Mean

## Definition

Harmonic Mean is an average that is useful for rates and ratios.

It gives more importance to smaller values compared with the arithmetic mean.

## Formula

Harmonic Mean = n / (1/x1 + 1/x2 + ... + 1/xn)

## Numerical Example

Consider:

30 and 60

Harmonic Mean:

= 2 / (1/30 + 1/60)

= 2 / (2/60 + 1/60)

= 2 / (3/60)

= 40

So, the harmonic mean is 40.

## Business Example

Harmonic mean can be used to calculate an average speed or average rate when the same distance is involved.

For example, a delivery vehicle travels at different speeds over equal distances.

## AI/ML Example

Harmonic mean is used in the F1-score.

F1-score combines Precision and Recall using their harmonic mean.

Formula:

F1 = 2 × (Precision × Recall) / (Precision + Recall)

This is useful when both precision and recall are important for evaluating a classification model.

```python
# Finding harmonic mean

import statistics

numbers = [30, 60]

harmonic_mean = statistics.harmonic_mean(numbers)

print("Harmonic Mean:", harmonic_mean)
```

```python
# Calculating F1 score using harmonic mean

precision = 0.80
recall = 0.60

f1_score = 2 * (precision * recall) / (precision + recall)

print("F1 Score:", f1_score)
```

# Comparison of Measures of Central Tendency

| Measure | Meaning | Common Use |
|---|---|---|
| Mean | Simple average | General numerical data |
| Median | Middle value | Data with outliers |
| Mode | Most frequent value | Categorical or repeated data |
| Weighted Mean | Average with importance | Scores and business calculations |
| Geometric Mean | Multiplicative average | Growth rates and ratios |
| Harmonic Mean | Average of rates | Speed, rates, and F1-score |

## Conclusion

Measures of central tendency help us understand the central or typical value of a dataset.

Mean is useful for normal numerical data.

Median is useful when the data contains extreme values.

Mode is useful for finding the most frequently occurring value.

Weighted Mean is useful when different values have different importance.

Geometric Mean is useful for growth rates and ratios.

Harmonic Mean is useful for rates and is also used in machine learning through the F1-score.
